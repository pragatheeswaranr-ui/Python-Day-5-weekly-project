```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../output/clean_data.csv")

df.head()
```

```python
print("Rows and Columns:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
df.describe()
```

```python
df['Attrition'].value_counts()
```

```python
sns.countplot(data=df, x='Attrition')

plt.title("Employee Attrition Count")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

plt.show()
```

```python
attrition_percentage = (
    df['Attrition'].value_counts(normalize=True) * 100
)

print(attrition_percentage)
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x='Department',
    hue='Attrition'
)

plt.title("Department-wise Employee Attrition")
plt.xticks(rotation=15)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Age'], bins=20, kde=True)

plt.title("Employee Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='Attrition',
    y='MonthlyIncome'
)

plt.title("Monthly Income vs Attrition")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x='OverTime',
    hue='Attrition'
)

plt.title("Overtime vs Employee Attrition")

plt.show()
```

```python
plt.figure(figsize=(15,10))

sns.heatmap(
    df.select_dtypes(include='number').corr(),
    annot=False,
    cmap='coolwarm'
)

plt.title("Employee Data Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(15,10))

sns.heatmap(
    df.select_dtypes(include='number').corr(),
    annot=False,
    cmap='coolwarm'
)

plt.title("Employee Data Correlation Heatmap")

plt.show()
```

```python
print("EDA FINDINGS")
print("----------------------")

print("1. Total Employees:", len(df))

print("\n2. Attrition Count:")
print(df['Attrition'].value_counts())

print("\n3. Department-wise Attrition:")
print(pd.crosstab(df['Department'], df['Attrition']))

print("\n4. Overtime-wise Attrition:")
print(pd.crosstab(df['OverTime'], df['Attrition']))

print("\n5. Average Monthly Income:")
print(df.groupby('Attrition')['MonthlyIncome'].mean())

print("\n6. Average Age:")
print(df.groupby('Attrition')['Age'].mean())
```

Exploratory Data Analysis – Findings

The dataset contains 1470 employee records.

Employee attrition was analysed using count plots.

Department-wise attrition was compared to identify differences between departments.

Age distribution was analysed to understand employee demographics.

Monthly income was compared between employees who stayed and left.

Overtime patterns were analysed to understand their relationship with attrition.

A correlation heatmap was used to examine relationships between numerical variables.
