
## 1. Bar Chart – Employee Count by Department

This chart shows the number of employees in each department.


```python

department_count = df["Department"].value_counts()

plt.figure(figsize=(8, 5))

department_count.plot(kind="bar", color="skyblue")

plt.title("Employee Count by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.legend(["Employees"])

plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig("reports/figures/bar_chart.png")

plt.show()
```


**Insight:** The bar chart shows the number of employees in each department. It helps identify which department has the highest and lowest number of employees.



## 2. Histogram – Employee Monthly Income Distribution

This histogram shows the distribution of monthly income among employees.


```python

plt.figure(figsize=(8, 5))

plt.hist(df["MonthlyIncome"], bins=15, color="orange",
         edgecolor="black")

plt.title("Distribution of Employee Monthly Income")
plt.xlabel("Monthly Income (Currency Units)")
plt.ylabel("Number of Employees")

plt.tight_layout()

plt.savefig("reports/figures/histogram.png")

plt.show()
```


**Insight:** The histogram shows the distribution of monthly income among employees. It helps identify the income ranges where most employees are grouped.



## 3. Scatter Plot – Age vs Monthly Income

This chart shows the relationship between employee age and monthly income.


```python

plt.figure(figsize=(8, 5))

plt.scatter(df["Age"], df["MonthlyIncome"],
            color="green", alpha=0.6)

plt.title("Employee Age vs Monthly Income")
plt.xlabel("Age (Years)")
plt.ylabel("Monthly Income (Currency Units)")

plt.tight_layout()

plt.savefig("reports/figures/scatter_plot.png")

plt.show()
```


**Insight:** The scatter plot shows the relationship between employee age and monthly income. It helps understand how income varies across different age groups.



## 4. Line Chart – Employee Count by Years at Company

This chart shows the number of employees based on their years of experience in the company.


```python

yearly_employees = df["YearsAtCompany"].value_counts().sort_index()

plt.figure(figsize=(8, 5))

plt.plot(yearly_employees.index, yearly_employees.values,
         marker="o", color="red", label="Employee Count")

plt.title("Employee Count by Years at Company")
plt.xlabel("Years at Company (Years)")
plt.ylabel("Number of Employees")

plt.legend()

plt.tight_layout()

plt.savefig("reports/figures/line_chart.png")

plt.show()
```


**Insight:** The line chart shows how employee count varies with years spent at the company. It helps identify the most common employee service periods.



## 5. Box Plot – Monthly Income by Department

This chart compares the monthly income distribution across different departments.


```python

plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="Department", y="MonthlyIncome",
            color="lightblue")

plt.title("Monthly Income Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Monthly Income (Currency Units)")

plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig("reports/figures/box_plot.png")

plt.show()
```
